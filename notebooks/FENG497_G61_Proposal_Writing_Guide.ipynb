## General rules for every section

1. **Use the official template.** Write in the draft in `04_Proposal`, in the section you are working on; use *Suggesting* mode when you change someone else's text. Keep the template's headings, numbering and fonts.
2. **One project, one story.** Every section uses the same title, the same numbers and the same words: *126 configurations*, *7 placements × 6 sampling rates × 3 channel sets*, *macro F1*, *leave-one-subject-out*, *battery life*, *cost per kit*, *Pareto front*. Copy them from the project notebook's Abstract and Project summary, not from memory.
3. **Specific beats general.** "The ankle accelerometer needs up to 100 Hz to keep heel-strike impacts" earns points; "sampling rate is important" does not. Take numbers from the project notebook's **Results log** (section 16) and its "What we found" notes.
4. **Write done work as a result, future work as a plan.** WP1–WP3 are done and WP5 has a first baseline result: describe what was done and show the numbers. WP4–WP7 are planned: say what will be done, how, and what the output will be.
5. **Show interdisciplinary integration.** The jury scores it. Name the four departments (aerospace, mechatronics, computer, industrial engineering) and what each contributes, at least in the Introduction, Objectives and Scope.
6. **Cite SDG 3 and SDG 9** by number and name in the Introduction, and refer back to them in the Objectives.
7. **Figures:** at most 3–4 in the whole proposal, each numbered, captioned and referred to in the text ("Figure 2 shows …"). Use the 300-dpi files the project notebook saves in `outputs/figures/`: the work-package flow, minutes per activity, required sampling rates, and the confusion matrix of the first baseline.
8. **Citations:** numbered [1], [2] … in order of first use, all from the shared Zotero library. Every claim taken from a paper gets a number; no copied sentences.
9. **Language:** short sentences, active voice, "we" for the group. No "very", "huge", "novel" without proof. Spell-check before uploading.

## 1. Abstract (10 points, 1 page max)

The abstract is the whole proposal in 200–300 words. Many readers read only this, so it must stand alone.

**Include, in this order (one or two sentences each):**

1. **Problem:** wearable activity monitors use several IMUs at high rates → expensive, power-hungry, uncomfortable.
2. **Question:** how few IMUs, where on the body, at what sampling rate, still recognise daily activities accurately?
3. **Method:** PAMAP2; 126 configurations (7 × 6 × 3) made from the same recordings; Random Forest, leave-one-subject-out, macro F1; power model; cost model.
4. **Expected result:** a Pareto front of accuracy, battery life and cost, and a recommended setup for 2–3 user profiles.
5. **Impact:** SDG 3 and SDG 9; four engineering disciplines.

**How:** start from the *Project summary* at the top of the project notebook and shorten it. Write it **last**, after every other section is final, so the numbers match.

**Avoid:** citations, figures, abbreviations without their full name, and results you do not have yet (write "we expect" or "the study will show").

## 2. Introduction (15 points, 2 pages max)

The introduction answers two questions: **what is the problem, and why does it matter?** It should make the reader want the answer before they reach the Objectives.

**Include:**

1. **Context (≈ 1 paragraph):** activity monitoring with wearables — elderly care and independent living, rehabilitation, fitness. Give one or two cited facts (for example ageing population, cost of care, adoption of wearables).
2. **The problem (≈ 1 paragraph):** research systems use many sensors at high rates; real users wear a device only if it is cheap, small and lasts days on a battery. Accuracy alone does not make a usable product.
3. **The gap (≈ 1 paragraph):** earlier studies optimise placement *or* sampling rate, usually for accuracy only. Few combine placement, rate and channels, and fewer add energy and cost. Point forward to the Literature Survey.
4. **This project (≈ 1 paragraph):** the research question, PAMAP2, 126 configurations, three outputs, Pareto front.
5. **Why it matters — SDGs (≈ 1 paragraph):** **SDG 3 Good Health and Well-being** (affordable monitoring for older adults and patients) and **SDG 9 Industry, Innovation and Infrastructure** (low-cost, energy-efficient health technology). Say *how* the project contributes, not only the numbers.
6. **Interdisciplinary team (2–3 sentences):** which department brings what.
7. **Structure of the proposal (1 sentence).**

**How:** go from wide to narrow — the world, the problem, the gap, our study. Each paragraph starts with its main point. One figure is welcome here: the work-package flow from the Week 2 deck.

**Avoid:** a history of wearables; method details (they belong in Scope); claims without a citation.

## 3. Objectives (15 points, ≈ ½ page)

The template asks for objectives that are **specific and measurable**: what will be *calculated*, *built* and *delivered*. A reader must be able to check at the end whether each one was met.

**Include:** one main objective (the research question in one sentence), then 5–7 numbered specific objectives. Each one names an output, a measure and, where possible, a target. A good pattern: *verb + what + how measured + by when*.

| # | Example objective | Output | WP |
| --- | --- | --- | --- |
| O1 | Build a reproducible preprocessing pipeline that turns PAMAP2 into 4 s feature windows for all 126 configurations | Feature tables at 6 rates (done) | WP2, WP3 |
| O2 | Measure recognition accuracy of every configuration as macro F1 under leave-one-subject-out validation | Accuracy table, 126 rows | WP5 |
| O3 | Estimate battery life (days) of every configuration from a datasheet-based power model | Energy table + equation | WP4 |
| O4 | Estimate the hardware cost of one kit for every configuration | Cost table + business case | WP7 |
| O5 | Identify the Pareto-optimal configurations and recommend one for each of 2–3 user profiles | Pareto plot + recommendation table | WP5, WP7 |
| O6 | Deliver an interactive dashboard where a user sets budget and battery life and sees the best setup | Working dashboard, open-source code | WP6 |

**How:** start each objective with an action verb (*build, measure, estimate, identify, deliver*). Tie each one to a work package, so the Gantt chart can follow it. If a target number is not known yet, give a range or say how it will be fixed (for example "the F1 threshold for each profile is set in WP7 from the literature").

**Avoid:** vague verbs (*investigate, explore, understand*); objectives that are really tasks (*read papers*); more objectives than you can finish by January.

## 4. Scope (15 points, ≈ 1 page)

The scope draws the borders of the project: **what is in, what is out, and what is fixed**. A clear scope protects the group from doing too much and tells the jury exactly what to expect.

**Include:**

1. **Data:** PAMAP2, 9 subjects, the 12 protocol activities; no new data collection. Mention that subject 109 is used only for training and why.
2. **The factor table** (copy from the project notebook): placements (7), sampling rates (6), channel sets (3) = 126 configurations.
3. **Outputs:** accuracy (macro F1, leave-one-subject-out), battery life (days), cost per kit, Pareto front, user-profile recommendations.
4. **Fixed choices** (from the project notebook's decision log): 4 s windows with 50 % overlap; ±16 g accelerometer; Random Forest baseline; sampling rates simulated by anti-aliased downsampling.
5. **Out of scope:** new hardware or recording; real-time on-device implementation; clinical validation; heart-rate and temperature sensors; fall detection (named as future work).
6. **Assumptions:** for example that downsampling 100 Hz data represents a sensor that records at the lower rate, and that datasheet currents represent real power use.

**How:** use one table for the factors and short bullets for in / out. One figure fits well here: the minutes-per-activity heatmap from the project notebook, to show what the data covers.

**Avoid:** method detail that belongs in a later report; a scope that does not match the Objectives (every objective must be inside the scope).

## 5. Literature Survey (20 points, 1–3 pages)

This is the section with the most points. The template asks you to **compare the content and method of earlier work with this study**. A list of paper summaries does not do that; a comparison organised by topic does.

**Include:**

1. **The dataset and benchmark:** the PAMAP2 paper and published results on it.
2. **The recognition pipeline:** windows, features, classifiers, and why subject-wise evaluation matters.
3. **Sensor placement:** which body positions and combinations earlier studies found best.
4. **Sampling rate:** how low the rate can go before accuracy drops.
5. **Orientation and sensor fusion:** complementary and Madgwick filters.
6. **Energy and cost:** studies that model power or cost of wearables (few exist — that is part of the gap).
7. **A comparison table** (the core of the section): one row per key study, columns *dataset · sensors and placement · sampling rate · method · evaluation · optimises for · main result*. The last row is **this study**, which is the only one covering placement + rate + channels + energy + cost.
8. **The gap (closing paragraph):** two or three sentences that the Objectives then answer.

**How:** at least 15 papers from the shared literature matrix, grouped under the topics above, not one paragraph per paper. For each topic: what is known, how it was done, what is missing. Start with the six starter papers from the Week 2 Pack, and add newer work (2023–2026) so the survey is current.

**Avoid:** copying abstracts; papers you have not read; any claim without its [number].

## 6. Risk Analysis (10 points, ≈ ½ page)

The template repeats the Gantt instructions in this section by mistake. **Write a risk register instead:** one row per risk, with likelihood, impact, how you reduce it, and who watches it. Aim for 6–8 real risks of *this* project.

| Risk | Likelihood | Impact | Mitigation | Owner |
| --- | --- | --- | --- | --- |
| PAMAP2 is small (9 subjects); results may not generalise | High | Medium | Leave-one-subject-out; report spread across folds; name a second dataset as future work | WP5 lead |
| Downsampled data differs from a sensor that records at a low rate | Medium | Medium | Anti-aliased downsampling; state the assumption in Scope | WP3 lead |
| Datasheet currents differ from real power use | Medium | Medium | Use 2–3 parts per component; report a range, not one number | WP4 lead |
| 126 × 8 model fits take too long in Colab | Medium | Low | Fewer trees first; save results after each configuration | WP5 + WP6 leads |
| Rare classes (rope jumping, running) give unstable F1 | High | Low | Macro F1 with per-class results; discuss in the report | WP5 lead |
| Prices change or are not public | Medium | Low | Distributor list prices with date; sensitivity range | WP7 lead |
| Member unavailable near deadlines (midterms, holidays) | Medium | Medium | Lead + support for every WP; weekly logs; draft by 5 Nov | All |
| Code lost or overwritten | Low | High | Shared Drive with copies; GitHub from Week 5–6 | WP6 lead |

**How:** rate likelihood and impact as Low / Medium / High; order the table from highest to lowest risk; make every mitigation an action someone can take.

**Avoid:** generic risks with no action ("the project may be late").

## 7. Gantt Chart (10 points, 1 page)

The template asks for **monthly columns from October 2026 to January 2027** and **tasks specific to this project**. The jury checks that the plan is realistic and that every objective has time in it.

**Include:** one row per work package (and the proposal and final-report writing), the lead's name on each row, the dependencies (WP5, WP4 and WP7 need WP3's outputs; the Pareto step needs all three), and the fixed milestones: proposal 13 Nov, midterm in Week 8, final report 8 Jan 2027, presentations and jury 15 Jan 2027.

The cell below draws a **suggested** timeline. It is a starting point, not a decision: agree the spans in the next meeting, change the dates in `TASKS` and `MILESTONES`, and run the cell again. *File → Download* or right-click the figure to save it as an image.

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from datetime import date

# Edit these after the meeting: (task, start, end, done?)
TASKS = [
    ("WP1 Literature review",       date(2026, 10, 8),  date(2026, 11, 5),  False),
    ("WP2–WP3 Fusion + features",   date(2026, 10, 15), date(2026, 10, 28), True),
    ("Proposal writing",            date(2026, 10, 22), date(2026, 11, 13), False),
    ("WP4 Energy model",            date(2026, 11, 2),  date(2026, 12, 4),  False),
    ("WP5 Classification",          date(2026, 11, 2),  date(2026, 12, 11), False),
    ("WP7 Cost + business case",    date(2026, 11, 9),  date(2026, 12, 11), False),
    ("Pareto + recommendations",    date(2026, 12, 7),  date(2026, 12, 25), False),
    ("WP6 Software + dashboard",    date(2026, 11, 16), date(2027, 1, 8),   False),
    ("Final report writing",        date(2026, 12, 14), date(2027, 1, 8),   False),
]
MILESTONES = [
    ("Proposal submitted",   date(2026, 11, 13)),
    ("Final report due",     date(2027, 1, 8)),
    ("Presentations + jury", date(2027, 1, 15)),
]

rows = [t[0] for t in TASKS] + [m[0] for m in MILESTONES]
fig, ax = plt.subplots(figsize=(11, 0.45 * len(rows) + 1.5))
for i, (name, start, end, done) in enumerate(TASKS):
    ax.barh(i, (end - start).days, left=mdates.date2num(start), height=0.5,
            color="#7BC47F" if done else "#C9D2E3", edgecolor="#5B6478")
for j, (name, d) in enumerate(MILESTONES):
    y = len(TASKS) + j
    first = j == 0
    ax.plot(mdates.date2num(d), y, "D", ms=9, color="#E8792B" if first else "#5B6478")
    ax.annotate(d.strftime("%d %b"), (mdates.date2num(d), y), xytext=(10, -4), textcoords="offset points",
                fontweight="bold" if first else "normal")
ax.axvline(mdates.date2num(MILESTONES[0][1]), color="#E8792B", ls="--", lw=1)

ax.set_yticks(range(len(rows)), rows); ax.invert_yaxis()
ax.xaxis_date(); ax.xaxis.set_major_locator(mdates.MonthLocator()); ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %Y"))
ax.set_xlim(mdates.date2num(date(2026, 10, 1)), mdates.date2num(date(2027, 1, 25)))
ax.grid(axis="x", alpha=0.3); ax.spines[["top", "right"]].set_visible(False)
ax.set_title("Proposal on 13 Nov, then WP4–WP7 run in parallel until the final report\n"
             "Suggested plan · green = already done", loc="left", fontsize=11)
plt.tight_layout(); plt.show()
```

**How:** build the final chart in the template's format, as a table or spreadsheet with months (or weeks) as columns and shaded cells for each task; mark milestones with a symbol. Use the same task names as the Objectives so the reader can match them. Add the lead's name to each row.

**Avoid:** generic tasks ("research", "coding"), tasks with no owner, and a plan that ends exactly on the deadline with no time for revision.

## 8. References (5 points)

**Include:** every source cited in the text, and nothing that is not cited. Numbered [1], [2] … in order of first appearance, one consistent style (IEEE fits an engineering proposal).

**Avoid:** Wikipedia, blogs and AI tools as sources; numbers typed by hand (they break when a citation moves).
